## 1. Soma dos Elementos da Lista
Dada uma lista de números, calcule a soma de todos os elementos.

```python
# Seu código aqui...
lista = [1, 2, 3, 4, 5]
soma =0
for i in lista:
  soma += i
print(f"A soma dos números dados na lista é de {soma}")
```
